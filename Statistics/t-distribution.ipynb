```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, t
from matplotlib import cm

x = np.linspace(-5,5,1000)

# dof to demonstrate covergence to normal distribution
dof = [1, 2, 5, 10, 30, 100]

# use inverted virdis colormap (virdis_r for reverse virdis)
cmap = cm.get_cmap('viridis_r', len(dof))

# set background color to white
plt.figure(figsize=(10, 6), facecolor='white')
ax = plt.gca()
ax.set_facecolor('white')

# plot normal distribution for comparison (mean=0, std=1)
plt.plot(x, norm.pdf(x), label='Standard Normal Distribution', color='black', linestyle='--', lw=2)

# plot the t-distributions for different degrees of freedom using the inverse color map
for i, df in enumerate(dof):
    plt.plot(x, t.pdf(x, df), label=f't-distribution (dof={df})', color=cmap(i), lw=2)

# set title and labels
plt.title('Convergence of t-distribution to Normal Distribution', fontsize=16)
plt.xlabel('x', fontsize=14)
plt.ylabel('PDF', fontsize=14)
plt.legend(facecolor='white', edgecolor='black',framealpha=1, labelcolor='black')

# adjust tick colour for the white color scheme
ax.tick_params(colors='black')

# show grid for better visibility
plt.grid(True)

# show the plot
plt.show()
```
